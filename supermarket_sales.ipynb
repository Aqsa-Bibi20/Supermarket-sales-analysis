```python
# 1. Total sales per Product Line calculate karein
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

# 2. Bar chart draw karein
plt.figure(figsize=(10, 5))
sns.barplot(x=product_sales.index, y=product_sales.values, palette='Blues_r')
plt.title('Total Revenue by Product Line', fontsize=14)
plt.xlabel('Product Line', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
# Payment methods count karein
payment_counts = df['Payment'].value_counts()

# Pie Chart draw karein
plt.figure(figsize=(6, 6))
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', colors=['#4CAF50', '#2196F3', '#FF9800'], startangle=140)
plt.title('Distribution of Payment Methods', fontsize=14)
plt.show()
```

```python
# 1. Time se Hour nikal kar naya column banayein
df['Hour'] = pd.to_datetime(df['Time'], format='%H:%M').dt.hour

# 2. Hourly sales calculate karein
hourly_sales = df.groupby('Hour')['Total'].sum()

# 3. Line chart draw karein
plt.figure(figsize=(9, 5))
plt.plot(hourly_sales.index, hourly_sales.values, marker='o', color='purple', linewidth=2.5)
plt.title('Sales Trend by Hour of Day (Peak Hours)', fontsize=14)
plt.xlabel('Hour of Day (24-Hour Format)', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Branch wise total gross income calculate karein
branch_income = df.groupby('Branch')['gross income'].sum().reset_index()

# Bar chart draw karein
plt.figure(figsize=(7, 5))
sns.barplot(x='Branch', y='gross income', data=branch_income, palette='viridis')
plt.title('Gross Income by Branch (A vs B vs C)', fontsize=14)
plt.xlabel('Supermarket Branch', fontsize=12)
plt.ylabel('Total Gross Income ($)', fontsize=12)
plt.show()
```

```python
# Gender wise Product Line sales count
plt.figure(figsize=(10, 5))
sns.countplot(x='Product line', hue='Gender', data=df, palette='coolwarm')
plt.title('Product Category Preference by Gender', fontsize=14)
plt.xlabel('Product Line', fontsize=12)
plt.ylabel('Number of Purchases', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Gender')
plt.show()
```

```python
# Average rating calculate karein har branch ke liye
avg_rating = df.groupby('Branch')['Rating'].mean().reset_index()

# Bar chart draw karein
plt.figure(figsize=(7, 4))
sns.barplot(x='Branch', y='Rating', data=avg_rating, palette='mako')
plt.title('Average Customer Rating by Branch', fontsize=14)
plt.xlabel('Supermarket Branch', fontsize=12)
plt.ylabel('Average Rating (out of 10)', fontsize=12)
plt.ylim(0, 10)
plt.show()
```
